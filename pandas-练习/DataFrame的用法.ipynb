```python
# 使用字典创建
import pandas as pd

data = {
    'Column1': [11, 22, 33],
    'Column2': ['A', 'B', 'C']
}
df = pd.DataFrame(data)
# print(df)
# # 
df
```

```python
# 访问列
column_data = df["Column1"]
column_data
```

```python
# 访问行
row_data = df.loc[0] # 通过标签访问
row_data
```

```python
# 访问行
row_data = df.iloc[0] # 通过 整数 位置访问
row_data
```

```python
# 同时访问 行和列
df.loc[0:2,'Column1':'Column2']  # 选择 <0-2> 行 与 <'Column1'-'Column2'> 列 交会的区域
```
