```python
# torch.chunk函数的使用
x = torch.arange(12).reshape(3, 4)
print(x)
```

```python
chunks = torch.chunk(x, 3, dim=0)  # 沿着第0维分成3块
print("分割后的块：")
for i, chunk in enumerate(chunks):
    print(f"Chunk {i}:\n{chunk}")
```

```python
# pytorch中的 contiguous() 的作用
"""
哪些操作会导致张量不连续：
    1. 索引操作
    2. 切片操作
    3. 转置操作
作用：
    contiguous() 方法的作用是将一个不连续的张量转换为一个连续的张量。
    它不会改变张量的形状或数据内容，只是重新排列内存中的数据，使得张量在内存中连续存储。
为什么需要 contiguous()：
    在 PyTorch 中，有些操作（如索引、切片、转置等）可能会导致张量的内存不连续，这可能会导致错误或性能问题
    这会影响到后续的操作。因此，在进行这些操作之前，需要使用 contiguous() 方法将张量转换为连续的张量。
    eg:
    view() 方法用于改变张量的形状，但它要求输入张量必须是连续的。如果张量是不连续的，
    直接调用 view() 会导致错误。在这种情况下，需要先调用 contiguous()。
"""

import torch

x = torch.tensor([[1, 2], [3, 4]])
print(x.is_contiguous())  # True

y = x.transpose(0, 1)
print(y.is_contiguous())  # False
```

## nn.ReLU() 中 参数 inplace=True/False 分别代表的含义!
1. inplace=True: 函数执行过程中，将输入的 Tensor 直接修改，并返回该 Tensor。
2. inplace=False: 函数执行过程中，将输入的 Tensor 修改，并返回一个修改后的 Tensor。
3. inplace=True/False 默认值都是 False。

##### 当 inplace=true时有什么优缺点呢？
a. 优点：
    减少内存使用，因为不需要额外的张量来存储结果。
    可以节省时间，因为不需要复制张量。
    节省内存，因为不需要额外的张量来存储结果。
    在内存受限的情况下非常有用。
b. 缺点：
    inplace 操作会改变原始张量，这可能会导致一些问题，例如，如果原始张量被其他地方使用，那么这些地方可能会收到不可expected的结果。
    inplace 操作可能会破坏梯度计算，因为原始张量被修改，这可能会导致梯度计算不正确。
    inplace 操作可能会导致内存泄漏，因为原始张量被修改，但内存不会被释放。
    inplace 操作可能会导致性能问题，因为原始张量被修改，这可能会导致内存使用增加，

我的看法：
   尽量使用 其默认值 inplace=False， 避免出现问题。不在乎这么点显存。

```python
# --------------------------------- 探究 nn.ReLU() 函数的参数  inplace=True 的反应 --------------------------------
import torch
import torch.nn as nn
import torch.nn.functional as F

x = torch.randn(3,5)
print('raw-x:',x)
x_relu = nn.ReLU(inplace=True)(x)   #  x 的原值直接被修改了
print('after-relu:',x_relu)  # 与上面的x相比，确实发生了变化 (这个司空见惯)
print('is_change_x:', x)     # x 也被原地修改了
print((x==x_relu).all().item())
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

x = torch.randn(3,5)
print('raw-x:',x)
x_relu = nn.ReLU(inplace=False)(x)   #  x 的原值直接被修改了
print('after-relu:',x_relu)  # 与上面的x相比，确实发生了变化 (这个司空见惯)
print('is_change_x:', x)     # x 本身不会发生变化!!!!
print((x==x_relu).all().item())
```
