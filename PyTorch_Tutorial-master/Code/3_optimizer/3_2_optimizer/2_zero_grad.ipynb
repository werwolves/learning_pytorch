```python
# coding: utf-8

import torch
import torch.optim as optim
import torch.nn as nn
# ----------------------------------- zero_grad

w1 = torch.randn(2, 2)
w1.requires_grad = True

w2 = torch.randn(2, 2)
w2.requires_grad = True
```

```python
w3 = torch.rand(2,3)
w3 = nn.Parameter(w3) 
w3.requires_grad
```

```python
id_w3 = id(w3)
print('id_w3:',id_w3,w3.requires_grad)
id_w3_data = id(w3.data)
print('id_w3_data:',id_w3_data,w3.data.requires_grad)
id_w3_detach = id(w3.detach())
print('id_w3_detach:',id_w3_detach,w3.detach().requires_grad)
```

```python
optimizer = optim.SGD([w1, w2], lr=0.001, momentum=0.9)

optimizer.param_groups[0]['params'][0].grad = torch.randn(2, 2)

print('参数w1的梯度：')
print(optimizer.param_groups[0]['params'][0].grad, '\n')  # 参数组，第一个参数(w1)的梯度
```

```python
optimizer.zero_grad()
print('执行zero_grad()之后，参数w1的梯度：')
print(optimizer.param_groups[0]['params'][0].grad)  # 参数组，第一个参数(w1)的梯度
```
