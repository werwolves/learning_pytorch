```python
from torch.optim.lr_scheduler import MultiStepLR
from torch.optim import Adam
import torch.nn as nn
from torch import optim
# 定义学习率调度器


class MyModel(nn.Module):
    def __init__(self):
        super(MyModel, self).__init__()
        self.pretrained_layers = nn.Sequential(
            nn.Linear(10, 20),
            nn.ReLU(),
            nn.Linear(20, 30)
        )
        self.new_layers = nn.Linear(30, 1)

    def forward(self, x):
        x = self.pretrained_layers(x)
        x = self.new_layers(x)
        return x

model = MyModel()

# 为预训练层和新层设置不同的学习率
optimizer = optim.Adam([
    {'params': model.pretrained_layers.parameters(), 'lr': 0.001},
    {'params': model.new_layers.parameters(), 'lr': 0.01}
])
### 学习率会在 epoch 30, 60, 90 时下降到之前的 0.1 倍
scheduler = MultiStepLR(optimizer, milestones=[30, 60, 90], gamma=0.1)
for i in range(100):
    # 假设每个epoch的学习率是0.1
    scheduler.step()
    print(f"Epoch {i+1}: Learning rate is {scheduler.get_last_lr()}")
```
