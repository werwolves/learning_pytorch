```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.optim.lr_scheduler import StepLR, MultiStepLR
```

```python
model = nn.Linear(2,1)
optimizer = optim.SGD(model.parameters(), lr=0.06)
```

```python
dir(optimizer)
```

```python
# dir(optimizer)
# optimizer.param_groups
# scheduler = StepLR(optimizer, step_size=3, gamma=0.1)  # 每隔step_size个epoch，lr=gamma*lr
scheduler = MultiStepLR(optimizer,milestones=[20,40],gamma=0.1) # 在epoch=milestones中的元素时， lr=gamma*lr
for epoch in range(10):
    optimizer.zero_grad()
    output = model(torch.randn(3,2))
    loss = output.pow(2).mean()
    loss.backward()
    optimizer.step()
    
    # 在每个epoch结束时调用调度程序
    print('-=' * 10, epoch, '=-' * 10)
    print('before-lr:',optimizer.param_groups[0]['lr'])
    scheduler.step()
    print('after-lr:',optimizer.param_groups[0]['lr'])
```
