## 1.普通的数据集构建

```python
import torch
from torch.utils.data import Dataset, DataLoader

class CustomDataset(Dataset):
    def __init__(self, data, labels):
        self.data = data
        self.labels = labels

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        sample = {'data': self.data[idx], 'label': self.labels[idx]}
        return sample

# Example data
data = torch.randn(100, 3, 224, 224)  # 100 samples of 3x224x224 images
labels = torch.randint(0, 2, (100,))  # 100 labels (binary classification)

# Create dataset
dataset = CustomDataset(data, labels)

# Create dataloader
dataloader = DataLoader(dataset, batch_size=4, shuffle=True)

# Iterate through the dataloader
for batch in dataloader:
    print(batch['data'].shape, batch['label'])
```

## 2.拼接数据集构建

```python
from torch.utils.data import ConcatDataset

# 假设我们有两个数据集 dataset1 和 dataset2
dataset1 = CustomDataset(data[:50], labels[:50])
dataset2 = CustomDataset(data[50:], labels[50:])

# 使用 ConcatDataset 拼接数据集
concat_dataset = ConcatDataset([dataset1, dataset2])

# 创建 dataloader
concat_dataloader = DataLoader(concat_dataset, batch_size=4, shuffle=True)

# 迭代拼接后的 dataloader
for batch in concat_dataloader:
    print(batch['data'].shape, batch['label'])
```
