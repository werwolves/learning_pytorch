```python
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import numpy as np
```

```python
cmap = cm.get_cmap('viridis')
cmap
```

```python
# 假设你有一个数据数组 data
# data = np.random.rand(10,10)
# data = np.random.rand(1,10)
data = np.arange(10).reshape(1,10)

plt.imshow(data, cmap=cmap)
plt.colorbar()  # 显示颜色条
plt.show()
```
