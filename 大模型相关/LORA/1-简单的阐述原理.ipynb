```python
import numpy as np
```

```python
x = np.random.rand(16, 100)
# 包含的参数量： 100 * 64 = 6400
w = np.random.rand(100, 64)   
# 包含的参数量： 100 * 4 + 4 * 64 = 1000 + 256 = 656
a = np.random.rand(100, 4)
b = np.random.rand(4, 64)
```

可见，LORA 只需要训练 656 个参数，而全量训练需要 6400 个参数。

```python
res = x @ w
res.shape
```

```python
res1 = x @ a @ b
res1.shape
```

```python
new_res = res + res1
new_res.shape
```
