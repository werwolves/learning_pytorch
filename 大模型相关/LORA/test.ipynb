```python
import random
def test():
    for _ in range(200):
        label = random.randint(0, 9)
        text = ' '.join(str(label) * 10)
        print(f"{label}\t{text}")
        yield {"text": text, "label": label}
```

```python
from datasets import Dataset
dataset = Dataset.from_generator(test)
dataset
```
