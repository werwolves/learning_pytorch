```python
# 阶乘
def jc(n):
    if n == 1:
        return 1
    return n * jc(n-1)
jc(5)
```

```python
# 猴子吃桃
def ct(day_num):
    if day_num == 10:
        return 1
    return (ct(day_num+1) + 1)*2
ct(1)
```
