```python
import torch
import torch.nn as nn
import torchvision
import torchvision.transforms as transforms
from torch.nn.utils import clip_grad_norm_
# Hyper-parameters
input_size = 28 * 28
num_classes = 10
num_epochs = 1
batch_size = 4
learing_rate = 0.001


# MNIST dataset (images and labels)
train_dataset = torchvision.datasets.MNIST(
    root='./data',
    train=True,
    transform=transforms.ToTensor(),
    download=False
)
val_dataset = torchvision.datasets.MNIST(
    root='./data',
    train=False,
    transform=transforms.ToTensor(),
)

train_loader = torch.utils.data.DataLoader(
    dataset=train_dataset,
    batch_size=batch_size,
    shuffle=True
)


test_loader = torch.utils.data.DataLoader(
    dataset=val_dataset,
    batch_size = batch_size,
    shuffle=False
)

# Logistic regression model
model = nn.Linear(input_size, num_classes)

# 
optimizer = torch.optim.Adam(model.parameters(),learing_rate)
criterion = nn.CrossEntropyLoss()

# Train the model
tot_steps = len(train_loader)
for epoch in range(num_epochs):
    for i, (images, labels) in enumerate(train_loader):
        # Reshape images to (batch_size, input_size)
        images = images.reshape(-1, input_size)
        model.train()
        # Forward pass
        outputs = model(images)
        loss = criterion(outputs, labels)
        
        optimizer.zero_grad()
        loss.backward()
        clip_grad_norm_(model.parameters(),0.5)
        optimizer.step()
        if i%100==0:
            print('Epoch [{}/{}], Step [{}/{}], Loss: {:.4}'
                  .format(epoch+1, num_epochs, i+1, tot_steps, loss.item()))
    # Test the model
    model.eval() # Dropout与BatchNorm都关掉or固定
    with torch.no_grad():# 禁用梯度计算，加速代码运行
        correct = 0
        total = 0
        for images, labels in test_loader:
            images = images.reshape(-1, input_size)
            outputs = model(images)
            # outputs.shape=[batch,class_num]
            _, predicts = torch.max(outputs,1)
            
            correct += torch.eq(predicts, labels).sum().item()
            total += labels.size(0)
        print('Accuracy of the model on testdataset:{} '.format(correct/total))
        
    # Save the model
    torch.save(model.state_dict(),'model.ckpt')
```
