```python
import torch.nn as nn
import torch
class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.conv_1 = nn.Conv2d(1,2,3)
        self.conv_2 = nn.Conv2d(2,2,3)
        self.fc = nn.Linear(2,3)
    def forward(self, x):
        x = self.conv_1(x)
        x = self.conv_2(x)
        x = self.fc(x)
        return x
model = Net()
```
