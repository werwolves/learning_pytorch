```python
import cv2
import numpy as np
im1 = cv2.imread('./imgs/0.jpg')
```

```python
gray = cv2.cvtColor(im1, cv2.COLOR_BGR2GRAY)
gray = np.float32(gray)
sift = cv2.SIFT_create()
# 计算SIFT特征
keypoints, descriptors = sift.detectAndCompute(im1, None)
# 在图像上绘制关键点
img_with_keypoints = cv2.drawKeypoints(im1, keypoints, None)

window_name = 'Image Window'  
cv2.namedWindow(window_name, cv2.WINDOW_NORMAL) 
cv2.resizeWindow(window_name, 2000, 1080) 
cv2.imshow(window_name,img_with_keypoints)
cv2.waitKey(0)
```

```python
"""
-------------- 关键点信息
在OpenCV中，keypoints 是一个包含关键点信息的列表，每个关键点是一个 cv2.KeyPoint 对象。cv2.KeyPoint 对象包含以下属性：
angle：关键点的方向，其值的范围是 [-180, 180] 或者 [-90, 0)。正值表示方向从x轴到关键点的角度，负值表示方向从y轴到关键点的角度。
class_id：分类标识符，用于聚类中的关键点。在大多数情况下，这个值是不变的。
octave：图像金字塔的层级，在该层级中检测到关键点。金字塔的每一层都对应于图像的特定尺度。
pt：关键点的位置，是一个包含浮点数 (x, y) 坐标的元组。
response：关键点的响应，即关键点的强度或突出程度。值越大，关键点越突出。
size：关键点的直径。
-------------- 描述向量
记录了关键点周围对其有贡献的像素点的一组向量值，其不受仿射变换，光照变换等影响
"""
# 遍历关键点
for kp,descrip in zip(keypoints,descriptors):
    # 打印每个关键点的信息
    print(f"Keypoint {kp.pt}:")
    print(f"  Angle: {kp.angle}")
    print(f"  Class ID: {kp.class_id}")
    print(f"  Octave: {kp.octave}")
    print(f"  Response: {kp.response}")
    print(f"  Size: {kp.size}")
    # 描述子的信息
    print(f"  descrip:{descrip.shape}\n") # 每个特征值为128维的向量!!!!
```
