```python
import cv2
import numpy as np
im1 = cv2.imread('./imgs/0.jpg')
im2 = cv2.imread('./imgs/1.jpg')
```

```python
# shi-Tomasi是Harris角点检测的改进
```

```python
gray = cv2.cvtColor(im1, cv2.COLOR_BGR2GRAY)
gray = np.float32(gray)
maxCorners = 1000  # 检测最大的角点数量
ql = 0.01      # 可接受的最小角点质量的阈值，低于这个阈值的角点被拒绝
minDistance = 10   # 表示返回的角点之间的最小可能的欧式距离
# shi-tomasi 角点检测
# 
# 
# 
corners = cv2.goodFeaturesToTrack(gray, maxCorners, ql, minDistance)
corners
```

```python
# harris角点的展示
# 对检测结果进行膨胀，以便标记角点
corners = np.intp(corners)
for corner in corners:
    print(f'corner:{corner}')
    x, y = corner.ravel()
    cv2.circle(im1, (x,y), 3, (255,0,0), -1)
    

window_name = 'Image Window'  
cv2.namedWindow(window_name, cv2.WINDOW_NORMAL) 
cv2.resizeWindow(window_name, 2000, 1080) 
cv2.imshow(window_name,im1)
cv2.waitKey(0)
```
