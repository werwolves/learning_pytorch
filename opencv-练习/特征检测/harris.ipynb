## 哈里斯角点检测!!

```python
import cv2
import numpy as np
im1 = cv2.imread('./imgs/0.jpg')
im2 = cv2.imread('./imgs/1.jpg')
```

```python
gray = cv2.cvtColor(im1, cv2.COLOR_BGR2GRAY)
gray = np.float32(gray)
blockSize = 3  # 检测窗口大小
ksize = 3      # Sobel卷积核大小
k = 0.04       # 权重系数，经验值， 一般取 0.02 ~ 0.04 之间
# harris角点检测
dst = cv2.cornerHarris(gray, blockSize, ksize, k)
dst
```

```python
dst.max()
```

```python
# harris角点的展示
# 对检测结果进行膨胀，以便标记角点
# dst = cv2.dilate(dst, None)
im1[dst>0.01*dst.max()] = [0,0,255]

window_name = 'Image Window'  
cv2.namedWindow(window_name, cv2.WINDOW_NORMAL) 
cv2.resizeWindow(window_name, 2000, 1080) 
cv2.imshow(window_name,im1)
cv2.waitKey(0)
```
