```python
import cv2
import numpy as np
im1 = cv2.imread('./imgs/0.jpg')
# 速度快，无版权!!， 效果不如sift与surf
```

```python
orb = cv2.ORB_create()
```

```python
gray = cv2.cvtColor(im1, cv2.COLOR_BGR2GRAY)
gray = np.float32(gray)
orb = cv2.ORB_create()
# 计算ORB特征
keypoints, descriptors = orb.detectAndCompute(im1, None)
# 在图像上绘制关键点
img_with_keypoints = cv2.drawKeypoints(im1, keypoints, None)

window_name = 'Image Window'  
cv2.namedWindow(window_name, cv2.WINDOW_NORMAL) 
cv2.resizeWindow(window_name, 2000, 1080) 
cv2.imshow(window_name,img_with_keypoints)
cv2.waitKey(0)
```
