```python
import cv2  
import numpy as np  
from shapely.geometry import Point, LineString, Polygon
def calculate_contour_area(contour):  
    try:  
        area = cv2.contourArea(contour)  
        return area  
    except Exception as e:  
        print("Error occurred:", e)  
        return None  

# 示例输入1：轮廓是一个多维数组  
contour1 = np.array([[10, 10], [10, 20], [20, 20], [20, 10]])  
# 方式一：计算多边形轮廓的面积
area1 = calculate_contour_area(contour1)  
print("Contour 1 area:", area1)  
# 方式二：计算多边形轮廓的面积
print("area=",Polygon(contour1).area)
```
