```python
import cv2  
import numpy as np  
  
# 创建一个空白图像  
img = np.zeros((300, 300, 3), dtype=np.uint8)  
  
# 定义多边形的顶点坐标  
# pts = np.array([[-100, 100], [300, 100], [250, 300], [150, 300]], dtype=np.int32)  
pts = np.array([[100, 100], [300, 100], [250, 300], [150, 300]], dtype=np.int32) 
  
# 填充多边形区域为红色  --- 以下3种写法效果相同!!!
# cv2.fillPoly(img, [pts], (0, 0, 255))  
# cv2.fillPoly(img, pts[None,:,:], (0, 0, 255))  
cv2.fillPoly(img, pts[np.newaxis,:,:], (0, 0, 255)) 

  
# 显示结果图像  
cv2.imshow("Filled Polygon", img)  
cv2.waitKey(0)  
cv2.destroyAllWindows()
```
