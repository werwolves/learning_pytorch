```python
import cv2  
import numpy as np  
  
# 旋转矩形的角度、中心点和大小  
angle = 45.0  # 旋转角度（度）  
width, height = 100, 50  # 矩形大小（宽，高）  
x, y = 50, 50  # 矩形中心点坐标（x，y）  
  
# 计算旋转矩形的四个角点坐标  
points = cv2.boxPoints(angle, (width, height), (x, y))  
  
# 打印角点坐标  
for point in points:  
    print(point)
```

```python
import cv2  
import numpy as np  
  
# 创建一个旋转矩形对象  
rect = (100, 50, 100, 50, 45)  # (x, y, width, height, angle)  
  
# 获取旋转矩形的四个角点坐标  
points = cv2.boxPoints(rect)  
  
# 打印角点坐标  
for point in points:  
    print(point)
```
