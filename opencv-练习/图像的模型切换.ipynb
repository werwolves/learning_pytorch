```python
import cv2
import os
from PIL import Image
import matplotlib.pyplot as plt
```

```python
img_path = r'./wq_text/wq_test_.png'
```

```python
im_BGR = cv2.imread(img_path)
# print(im_BGR)
im_YUV = cv2.cvtColor(im_BGR, cv2.COLOR_BGR2YUV)
cv2.imwrite('./wq_text/wq_test_yuv.png', im_YUV)
```

```python
def show_merge_img(im1, im2, label='add-title',im1_name='im1',im2_name='im2', figsize=(10,3)):
    fig, axs = plt.subplots(nrows=1, ncols=2, figsize=figsize)  
  
    # 在第一个子图上显示第一张图片  
    axs[0].imshow(im1)  
    axs[0].set_title(im1_name)  
    axs[0].axis('off')  
    
    # 在第二个子图上显示第二张图片  
    axs[1].imshow(im2)  
    axs[1].set_title(im2_name)  
    axs[1].axis('off')  

    plt.suptitle(label)  
    # 显示图像
    plt.show() 
```

```python
show_merge_img(im_BGR,im_YUV)
```
