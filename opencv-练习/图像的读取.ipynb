```python
import os,cv2
import numpy as np
```

```python
img_path = r"imgs 中文路径/1.png"
os.path.exists(img_path)
```

## 

```python
with open(img_path, 'rb') as file:
    img = file.read()    # 获取字节流
img = np.frombuffer(img, dtype="uint8") # 将字节流转换为 Numpy 数组
img.shape
```

```python
# 如果需要忽略方向信息，则解码图像时忽略EXIF方向信息，并以彩色模式 !!!!解码!!!! -----> 
img = cv2.imdecode(img, cv2.IMREAD_IGNORE_ORIENTATION | cv2.IMREAD_COLOR)
img.shape
```
