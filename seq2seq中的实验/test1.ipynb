```python
import torch
import torch.nn as nn
```

```python
# nn.Embedding()的主要作用是将整数索引
#             |
#            \|/
#          密集向量
# 
```

```python
# 假设词汇表大小 1000，
# 每个词汇表维度 50
embedding_dim = 50
vocab_size = 1000
embedding = nn.Embedding(vocab_size,embedding_dim)
# 索引为1的word对应的密集变量是 
# embedding(torch.tensor(1)).shape # torch.Size([50])
```
