```python
from torchvision.ops import DeformConv2d
# 使用的环境是 yolo-pytorch
import torch.nn as nn
import torch
```

```python
deform_conv = DeformConv2d(3,6,kernel_size=3, stride=1,padding=1)
conv = nn.Conv2d(3,6,kernel_size=3, stride=1,padding=1)
```

```python
input_torch = torch.randn(12,3,224,224)

offset = torch.randn((12, 2 * 3 * 3, 224, 224),requires_grad=True)  # 2*3*3是因为每个位置有x和y两个偏移量  
out1 = deform_conv(input_torch,offset)
out2 = conv(input_torch)
print(f'out1.shape:{out1.shape}')
print(f'out2.shape:{out2.shape}')
```
