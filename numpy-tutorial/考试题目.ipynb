```python
arr = np.array([10, 20, 30, 40, 50])
arr/arr.sum()
```

```python
def softmax(a):
    return np.exp(a)/np.exp(a).sum()
softmax(arr)
```

```python
def softmax(A):
    exp_A = np.exp(A)
    return exp_A / np.sum(exp_A)
softmax(arr)
```
