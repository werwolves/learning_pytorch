```python
#### 手动写一个函数求一个方阵的对角线（向上或下）上元素的索引
def diag_index(n, direction='up'):
    """ 获取对角线向 右上或 左下 放下移动一个位置后的元素索引 """
    if direction == 'up': # 向右上方移动一个位置
        return [(i, i+1) for i in range(n-1) if i+1 < n ]
    elif direction == 'down': # 向左下方移动一个位置
        return [(i, i-1) for i in range(1,n) if i < n ]
    elif direction == 'nomove': # 不移动
        return [(i, i) for i in range(n)]
    else:
        assert False, "direction must be 'up' or 'down' or 'nomove'"  
res = diag_index(4,'up') 
res
```

```python
res = diag_index(4,direction='down') 
res
```

```python
res = diag_index(4,direction='nomove') 
res
```
