```python
#### 手动写一个函数求一个方阵的对角线（向上或下）上元素的索引
def diag_index(n, direction='up'):
    """ 1. nomove: 获取对角线位置的索引
        2. up: 获取对角线向右上一个位置的索引 
        3. down: 获取对角线向左下一个位置的索引
        4. 其他: 获取去除1，2，3以外的索引
        右上或 左下 放下移动一个位置后的元素索引 """
    if direction == 'up': # 向右上方移动一个位置
        return [(i, i+1) for i in range(n-1) if i+1 < n ]
    elif direction == 'down': # 向左下方移动一个位置
        return [(i, i-1) for i in range(1,n) if i < n ]
    elif direction == 'nomove': # 不移动
        return [(i, i) for i in range(n)]
    else:
        special_loc = diag_index(n, direction='up') + diag_index(n, direction='down') +  diag_index(n, direction='nomove') 
        full_n_square = [(i,j) for i in range(n) for j in range(n)]
        return list(set(full_n_square) - set(special_loc))
res = diag_index(4,'up') 
res
```

```python
# 一次性生成所有的索引， 简单高效
def diag_index(n):
    """ 1. nomove: 获取对角线位置的索引
        2. up: 获取对角线向右上一个位置的索引 
        3. down: 获取对角线向左下一个位置的索引
        4. 其他: 获取去除1，2，3以外的索引
        右上或 左下 放下移动一个位置后的元素索引 """
    return_loc_dict = {
        "up": [(i, i+1) for i in range(n-1) if i+1 < n ],
        "down": [(i, i-1) for i in range(1,n) if i < n ],
        "nomove": [(i, i) for i in range(n)],
    }    
    return_loc_dict["other"] = [(i,j) for i in range(n) for j in range(n) if (i,j) not in return_loc_dict["up"] + return_loc_dict["down"] + return_loc_dict["nomove"]] 
    
    return return_loc_dict
res = diag_index(4)
```
