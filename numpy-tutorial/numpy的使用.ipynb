<span style='color:red;'>1. 生成确定数值的数组</span>

```python

np.linspace(start=1, stop=100, num=12) # 等差数列
a = np.zeros([12,4]) # 生成全 0 数组
np.zeros_like(a) # 生成 形状和a一样且元素全是0的数组
np.ones(10, dtype=np.int) # 同理 
# 生成单位矩阵
np.eye(3)
```

<span style='color:red;'>2.生成随机数组</span>

```python
# 0-1均匀分布
print('-='*5, '0-1均匀分布' ,'-='*5)
print('随机生成1个数：',np.random.rand())
print('随机生成3个数：',np.random.rand(3))
print('随机生成3*2个数：',np.random.rand(3,2))
# 0-1正态分布
print('-='*5, '0-1正态分布' ,'-='*5)
print('随机生成1个数：',np.random.randn())
print('随机生成3个数：',np.random.randn(3))
print('随机生成3*2个数：',np.random.randn(3,2))
# 指定范围的整数
print('-='*5, '随机生成指定范围的整数' ,'-='*5)
print('随机生成10个指定范围的整数',np.random.randint(low=0,high=10,size=10))
print('随机生成2*3个指定范围的整数',np.random.randint(low=0,high=10,size=(2,3)))
```

```python
a = np.arange(10)
print('a:',a)
np.random.shuffle(a) # 注意： 可见打乱的是原数组
print('shuffle_a:',a)
```

```python
arr = np.arange(16).reshape(4,4)
print('arr:  ', arr)
print('arr.T:', arr.T)
print('arr.T:', arr.transpose(1,0))
print('arr.T:', arr.transpose())
```

<span style='color:red;'> 3.数组的合并</span>

```python
arr1 = np.array([[1,2,3,4],[5,6,7,8]])
arr2 = np.array([[11,22,33,44],[55,66,77,88]])
print('concat-axis=0',np.concatenate([arr1,arr2], axis=0))
print('concat-axis=1',np.concatenate([arr1,arr2], axis=1))
```

```python
a = np.array([1,2,3])
np.diag(a) # 当是一维数组时， 结果形成以一维数组为对角线元素的矩阵
```

```python
a = np.arange(1,10).reshape(3,3)
print(a)
b = np.diag(a) # 当是二维数组时，结果输出对角线上的元素
b
```
