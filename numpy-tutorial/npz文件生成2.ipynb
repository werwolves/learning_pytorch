```python
import numpy as np
```

```python
# 可用于分割网络数据的生成
data_dict = {
    "img": np.array([1,2,3]),
    "label": np.array([4,5,6])
}
np.savez("data.npz", **data_dict)
```

```python
# 可用于分割网络数据的读取
data = np.load('data.npz')
img = data["img"]
label = data["label"]
data.close() # 记得关闭文件
```
