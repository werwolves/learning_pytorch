```python
import numpy as np

def batch_check_intersect(boxes_a: np.ndarray, boxes_b: np.ndarray):
    # boxes_a: [N,4], boxes_b:[M,4]
    N = boxes_a.shape[0]
    M = boxes_b.shape[0]
    a_x0, a_y0, a_x1, a_y1 = boxes_a.T
    b_x0, b_y0, b_x1, b_y1 = boxes_b.T

    # 广播计算所有配对相交坐标
    inter_x0 = np.maximum(a_x0[:, None], b_x0[None, :])
    inter_y0 = np.maximum(a_y0[:, None], b_y0[None, :])
    inter_x1 = np.minimum(a_x1[:, None], b_x1[None, :])
    inter_y1 = np.minimum(a_y1[:, None], b_y1[None, :])

    inter_w = inter_x1 - inter_x0
    inter_h = inter_y1 - inter_y0
    mask = (inter_w > 0) & (inter_h > 0)

    # 取出所有相交坐标对 (a_idx, b_idx)
    a_idx, b_idx = np.where(mask)
    return list(zip(a_idx, b_idx))

# 使用示例
boxes_a_np = np.array([[10,10,50,50],[60,60,100,100],[200,200,300,300]])
boxes_b_np = np.array([[30,30,70,70],[120,120,180,180]])
res = batch_check_intersect(boxes_a_np, boxes_b_np)
print(res)
```
