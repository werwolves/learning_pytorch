```python
import numpy as np  
  
x = np.array([1, 1, 2, 1, 2, 100]) # 统计range(0-max(x))从中各个元素出现的次数 
print(np.bincount(x))  # 输出：[0出现次数=0,1出现次数=3,2出现次数=3]  
print(np.argmax(np.bincount(x)))
```

```python
import numpy as np
labels, counts = np.unique(x, return_counts=True)  # 分类计数
print(labels) # [1 2 3]
print(counts) # [3 2 1]
print(labels[np.argmax(counts)])
```
