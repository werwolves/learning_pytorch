```python
import numpy as np

# 创建一些数组
arr1 = np.ones((4,3,3), dtype=object)
arr2 = np.ones((4,4,4), dtype=object)
arr3 = np.array([1,2,3,4], dtype=object)

# 定义 arr4
arr4 = [(np.array([3, 0], dtype=np.int64), 0.8, None), (np.array([2, 0], dtype=np.int64), 1.0, None)]

# 将 arr4 转换为结构化数组
coords = np.array([x[0] for x in arr4])
values = np.array([x[1] for x in arr4])
print(f"coords:{coords}")
print(f"values:{values}")
# 保存数组到 .npz 文件
np.savez('arrays.npz', arr1=arr1, arr2=arr2, arr3=arr3, coords=coords, values=values)
```
