```python
a = np.random.randint(0,10,(2,2))
b = np.random.randint(0,10,(2,2))
print('-'*6)
print(a)
print('-'*6)
print(b)
```

```python
# 这个一般用在哪里呢？！！！
np.fmax(a,b)  # 从a，b的对应位置处找到最大的值
```

```python
np.where(a>b,a,b) # 有条件又有对应条件的值，则返回对应条件的值
```
