```python
def longest_increasing_subsequence(arr):
    if not arr:
        return 0

    n = len(arr)
    dp = [1] * n  # 初始化 dp 数组，每个元素的最小上升子序列长度为 1

    for i in range(1, n):
        for j in range(i):
            if arr[i] > arr[j]:
                dp[i] = max(dp[i], dp[j] + 1)

    # max_length = max(dp)  # 直接从 dp 数组中找到最大值
    return dp  # 返回最长上升子序列的长度

# 示例数组
arr = [1, 8, 2, 3, 4, 5]
print(longest_increasing_subsequence(arr))
```
