```python
x = np.arange(4)
y = np.arange(4)
print('x:',x)
print('y:',y)
xx, yy = np.meshgrid(x,y)
print('xx:',xx)
print('yy:',yy)
xx_yy = np.concatenate([xx,yy],axis=-1)
print('xx_yy:',xx_yy)
```

```python
import numpy as np
# 方法一：从指定的列表中随机选择3个元素(又放回-里面可能有重复的)
sample = np.random.choice([1,2,3,4,5],size=3)
print(sample)
# 方式二：从0~9内选择5个不重复的元素
sample=np.random.choice(10,size=5,replace=False)
print(sample)
# 方式三：每个元素选择都指定对应的概率
sample=np.random.choice(3,size=2,p=[0.2,0.5,0.3],replace=False)
print(sample)
```
