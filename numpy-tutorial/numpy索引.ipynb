```python
import numpy as np
from random import shuffle,seed
```

```python
# 对一个数组的一种统一打乱方式   多组数据统一打乱， 还一一对应的打乱方式
a = np.array(['a','b','c','d','e','f','g'],dtype=object)
b = np.array(['aa','bb','cc','dd','ee','ff','gg'],dtype=object)
print("-"*5,'打乱之前的数据','-'*5)
print(f'a:{a}')
print(f'b:{b}')
print("-"*5,'打乱之后的数据','-'*5)
shuffle_index =  np.arange(len(a))
# seed(10100)
shuffle(shuffle_index)
a = a[shuffle_index]
b = b[shuffle_index]
print(f'shuffle_index:{shuffle_index}')
print(f'a:{a}')
print(f'b:{b}')
```
