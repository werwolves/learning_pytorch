```python
import numpy as np
from random import shuffle,seed
```

```python
# 对一个数组的一种统一打乱方式   多组数据统一打乱， 还一一对应的打乱方式
a = np.array(['a','b','c','d','e','f','g'],dtype=object)
b = np.array(['aa','bb','cc','dd','ee','ff','gg'],dtype=object)
print("-"*5,'打乱之前的数据','-'*5)
print(f'a:{a}')
print(f'b:{b}')
print("-"*5,'打乱之后的数据','-'*5)
shuffle_index =  np.arange(len(a))
# seed(10100)
shuffle(shuffle_index)
a = a[shuffle_index]
b = b[shuffle_index]
print(f'shuffle_index:{shuffle_index}')
print(f'a:{a}')
print(f'b:{b}')
```

```python
a[[1,0,4]]  # 这种 挑选筛选数据的方式 才是上述代码的精髓
```

```python
a[[1,2],0]
```

```python
a[1,2] # 1行2列
a[1:,2:] # [[6,7],[10,11]]
a[[1,2]] # 获取第1，2行的所有元素
# 总结： 其实可以将 [1,2]看成是一个整体，以','作为划分不同维度的标志
```
