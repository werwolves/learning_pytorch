```python
import numpy as np
```

```python
a = np.arange(12).reshape(3,4)
a
```

```python
a[[1,2],0]
```

```python
a[1,2] # 1行2列
a[1:,2:] # [[6,7],[10,11]]
a[[1,2]] # 获取第1，2行的所有元素
# 总结： 其实可以将 [1,2]看成是一个整体，以','作为划分不同维度的标志
```

```python
a = np.arange(10)
np.where(a>5)
a[a>5]
np.any(a>5) # True
np.all(a>5)  # False
```
