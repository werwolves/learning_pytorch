```python
import h5py
import numpy as np

# ----------------------------------------------------------------------- 保存对应的数据 --------------------------------------------------------------
# 创建一些示例数据
data = np.random.rand(100, 100)  # 100x100 的随机数组
labels = np.random.randint(0, 2, size=100)  # 100个随机的二分类标签

# 指定HDF5文件路径
file_path = 'example_data.h5'

# 以写入模式打开HDF5文件
with h5py.File(file_path, 'w') as hdf5_file:
    # 在HDF5文件中创建一个数据集，并存储数据
    hdf5_file.create_dataset('image_data', data=data)
    
    # 创建另一个数据集存储标签
    hdf5_file.create_dataset('image_labels', data=labels)
    
    # 可以添加更多的数据集或属性
    hdf5_file.attrs['description'] = 'Example dataset for machine learning'
```

```python
# ---------------------------------------------------------------------- 读取对应的数据 ------------------------------------------------------------------
import h5py

# 指定HDF5文件路径
file_path = 'example_data.h5'

# 以读取模式打开HDF5文件
with h5py.File(file_path, 'r') as hdf5_file:
    # 读取数据集
    image_data = hdf5_file['image_data'][:]
    image_labels = hdf5_file['image_labels'][:]
    
    # 读取文件属性
    description = hdf5_file.attrs['description']
    
    # 打印读取的数据和属性
    print("Image Data Shape:", image_data.shape)
    print("Image Labels Shape:", image_labels.shape)
    print("Description:", description)



```
