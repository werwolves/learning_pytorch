```python
# topk的使用
import numpy as np
import torch
t = np.arange(18)
np.random.shuffle(t)
t = torch.tensor(t)
top_value, top_index = t.topk(3)
print(f'top_value:{top_value}')
print(f'top_index:{top_index}')
```
