```python
import matplotlib
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = (4, 1) # 设置默认的画布大小为宽：4英寸，高1英寸
```

<span style='color:red;'>1.多子图显示一</span>

```python
x = np.linspace(1,10,100)
y = np.sin(x)
y1 = np.cos(x)

fig = plt.figure(figsize=(8,5)) # 申请一张画布
# fig = plt.figure()
img = Image.open('./1.jpg')

ax1 = fig.add_subplot(121)
# ax1.plot(x,y,'r--')\
ax1.imshow(img)
ax1.set_title('test1')

ax2 = fig.add_subplot(122)
ax2.plot(x,y1,'b-o')
ax2.set_title('test2')

plt.tight_layout() # 用于自动调整子图之间的空间，以便更好地利用整个figure区域。
plt.show()
```

<span style='color:red;'>2.多子图显示二</span>

```python
x = np.linspace(1,10,100)
y = np.sin(x)
y1 = np.cos(x)

# fig = plt.figure(figsize=(18,1)) # 申请一张画布

fig, axs = plt.subplots(1,2)

# 在每个axes上绘图
axs[0].plot(x,y,'r--')
axs[0].set_title('test1')

axs[1].plot(x,y1,'b-')
axs[1].set_title('test2')

plt.tight_layout()
plt.show()
```
