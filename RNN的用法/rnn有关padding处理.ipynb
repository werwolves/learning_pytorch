```python
import torch
import torch.nn as nn
```

```python
gru=nn.GRU(
    input_size=1,
    hidden_size=1,
    batch_first=True
)
input=torch.tensor([
    [1,2,3,4,5],
    [1,2,3,4,0],
    [1,2,3,0,0],
    [1,2,0,0,0]
]).unsqueeze(2) # shape=[4,5,1]
input_lengths=torch.tensor([5,4,3,2])
input=nn.utils.rnn.pack_padded_sequence(
    input,
    input_lengths,
    batch_first=True,
    enforce_sorted=False
)
# print(input)
output,hidden=gru(input.float())
print(output[0].shape) # torch.Size([14, 1])
output,_=nn.utils.rnn.pad_packed_sequence(
    sequence=output,batch_first=True
)
print(output)# 输入为0的区域，对应的输入也为0 
             # 普通的rnn没有这个功能
```
