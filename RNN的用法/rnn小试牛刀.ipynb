```python
import torch  
import torch.nn as nn  
  
# 定义 LSTM 网络  
class LSTMNet(nn.Module):  
    def __init__(self, input_size, hidden_size, output_size):  
        super(LSTMNet, self).__init__()  
        # input = [batch, seq_len, input_size]
        self.lstm = nn.LSTM(input_size, hidden_size)  
        # output = [batch, seq_len, hidden_size] # 
        
        self.fc = nn.Linear(hidden_size, output_size)  
  
    def forward(self, x):  
        # LSTM 的输出有两个部分：输出和隐藏状态  
        out, _ = self.lstm(x)  # out: [batch_size, sequence_length, hidden_size]  
        # out = self.fc(out[:, -1, :])  # 取最后一个时间步的输出作为预测  
        return out  
  
# 实例化网络  
input_size = 10  
hidden_size = 200  
output_size = 1  
net = LSTMNet(input_size, hidden_size, output_size)
#               batch,seq_len,    input_size
input = torch.ones(32,    100,    10).float()
out = net(input)
print(f'out-shape:{out.shape}') # [32, 1]
```
