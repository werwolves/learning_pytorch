```python
import torch
import torch.nn as nn
```

```python
input_size = 10
hidden_size = 20
batch_size = 32
seq_len = 6
num_layers = 2
bidirectional_flag = True

lstm = nn.LSTM(input_size=input_size, # 输入数据特征数
        hidden_size=hidden_size, # 隐藏层单元数
        num_layers=num_layers,  # lstm的层数
        bias=False,
        batch_first=True,
        bidirectional=bidirectional_flag)


input_data = torch.randn(batch_size, seq_len, input_size)
if bidirectional_flag:
        h0 = torch.zeros(num_layers*2,batch_size,hidden_size)
        c0 = torch.zeros(num_layers*2,batch_size,hidden_size)  
else:
        h0 = torch.zeros(num_layers,batch_size,hidden_size)
        c0 = torch.zeros(num_layers,batch_size,hidden_size)
# h0 = torch.zeros(batch_size,5,hidden_size)
# c0 = torch.zeros(batch_size,5,hidden_size)  
# out,hidden_out = lstm(input_data,(h0,c0)) #  torch.Size([4, 5, 20])
out,hidden_out = lstm(input_data) 
print('out.shape:',out.shape)
print('hidden_out:',hidden_out[0].shape, hidden_out[1].shape)
# hidden_out[0].shape # torch.Size([4, 5, 20])
# hidden_out[1].shape # torch.Size([4, 5, 20])
```
