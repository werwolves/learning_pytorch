```python
import torch

# 假设我们有以下特征向量
text_features = torch.randn(1, 300)  # 文字特征向量，维度为 (1, 300)
coord_features = torch.randn(1, 4)  # 坐标特征向量，维度为 (1, 4)

# 直接拼接
concatenated_features = torch.cat((text_features, coord_features), dim=1)  # 沿着维度1拼接

print("拼接后的特征向量维度:", concatenated_features.shape)
# 输出：拼接后的特征向量维度: torch.Size([1, 304])
```
