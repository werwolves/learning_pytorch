```python
import torch
import torch.nn as nn
import torch.nn.functional as F

# 假设我们有以下特征向量
text_features = torch.randn(1, 300)  # 文字特征向量，维度为 (1, 300)
coord_features = torch.randn(1, 4)  # 坐标特征向量，维度为 (1, 4)

# 定义一个简单的注意力模型
class SimpleAttention(nn.Module):
    def __init__(self, text_dim, coord_dim):
        super(SimpleAttention, self).__init__()
        # 定义一个线性层来计算权重
        self.attention_layer = nn.Linear(text_dim + coord_dim, 2)  # 输出两个权重，分别对应文字特征和坐标特征

    def forward(self, text_features, coord_features):
        # 拼接特征向量  ===> 维度为 (1, 304)
        combined_features = torch.cat((text_features, coord_features), dim=1)
        
        # 计算权重
        weights = self.attention_layer(combined_features)  # 权重维度为 (1, 2)
        weights = F.softmax(weights, dim=1)  # 使用softmax将权重归一化到0-1之间，使其和为1
        
        # 分离权重  text_weight.shape = (1, 1), coord_weight.shape = (1, 1)
        text_weight, coord_weight = weights[:, 0].unsqueeze(1), weights[:, 1].unsqueeze(1)
        
        # 加权融合
        # 加权融合：保留加权后的完整特征向量
        weighted_text_features = text_weight * text_features
        weighted_coord_features = coord_weight * coord_features
        weighted_features = torch.cat((weighted_text_features, weighted_coord_features), dim=1)
        return weighted_features

# 初始化注意力模型
attention_model = SimpleAttention(text_dim=300, coord_dim=4)

# 使用注意力模型进行加权融合
weighted_features = attention_model(text_features, coord_features)

print("加权融合后的特征向量维度:", weighted_features.shape)
# 输出：加权融合后的特征向量维度: torch.Size([1, 304])
```
