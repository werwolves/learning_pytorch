```python
from tensorboard import program
tb = program.TensorBoard()
tb.configure(argv=[None, "--logdir=tensorboard-log-1", "--port=6006"])
tb.launch()  # 此时不会再报pkg_resources错误
```

```python
# 补充写日志的代码（验证可视化）
from torch.utils.tensorboard import SummaryWriter
writer = SummaryWriter("tensorboard-log-1")
writer.add_scalar("test/loss", 0.8, global_step=1)
writer.add_scalar("test/acc", 0.9, global_step=1)
writer.close()
```
