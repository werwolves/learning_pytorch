```python
import torch
import torch.nn as nn
```

```python
# nn.ModuleList 是 PyTorch 中的一个容器类，用于存储多个神经网络模块（nn.Module 的子类）。
# 这个容器会自动将每个模块的参数（parameters）添加到网络中，并且允许你使用 
# 需要你在自己的模型中定义 forward 方法来手动调用这些模块
# ---------------------------------------------- #
# Python 列表的常规操作（如 extend 和 append）来管理这些模块。
# 内部已经实现了forward 方法，你可以直接传递数据给整个模型
```

```python
class MyModel(nn.Module):
    def __init__(self):
        super(MyModel,self).__init__():
        self.layers = nn.ModuleList([
            nn.Linear(10,20),
            nn.ReLU(),
            nn.Linear(20,30)
        ])
    def forward(self,x):
        for layer in self.layers:
            x = layers(x)
        return x
    # 并且可以通过append和extend动态的删除或添加模块
    def add_layer(self):
       self.layers.append(nn.Linear(30,40)) 
```
